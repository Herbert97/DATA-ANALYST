# 4 investigacion de relacion entre valores faltantes.

```python
# ===================================================================
# INVESTIGACIÓN DE RELACIONES ENTRE VALORES FALTANTES
# ===================================================================

print(" INVESTIGANDO RELACIONES ENTRE VALORES FALTANTES")
print("=" * 55)

# 1. ¿Hay relación entre 'internal' y 'operator_id' faltantes?
print("\n RELACIÓN ENTRE VALORES FALTANTES:")
print("\nTabla cruzada de valores faltantes:")
crosstab = pd.crosstab(
    df1['internal'].isnull(), 
    df1['operator_id'].isnull(), 
    margins=True
)
print(crosstab)

print("\n" + "-"*40)

# 2. Analizar por 'direction'
print("\n VALORES FALTANTES POR DIRECCIÓN:")
print("\nValores faltantes en 'operator_id' por dirección:")
direction_analysis = df1.groupby('direction').agg({
    'operator_id': lambda x: x.isnull().sum(),
    'internal': lambda x: x.isnull().sum()
}).rename(columns={'operator_id': 'operator_id_faltantes', 'internal': 'internal_faltantes'})
print(direction_analysis)

print("\n" + "-"*40)

# 3. ¿Qué valores tiene 'internal' cuando 'operator_id' está faltante?
print("\n VALORES DE 'internal' CUANDO 'operator_id' FALTA:")
internal_when_operator_missing = df1[df1['operator_id'].isnull()]['internal'].value_counts(dropna=False)
print(internal_when_operator_missing)
```

```python
# ===================================================================
# VERIFICACIÓN DETALLADA DE HIPÓTESIS: VALORES FALTANTES
# ===================================================================

print(" VERIFICACIÓN DE HIPÓTESIS: ¿Son todas llamadas internas?")
print("=" * 65)

# 1. Análisis detallado de los 8,172 valores faltantes en operator_id
print("\n DESGLOSE COMPLETO DE VALORES FALTANTES EN 'operator_id':")
print(f"Total de valores faltantes en operator_id: {df1['operator_id'].isnull().sum()}")

# Analizar qué valores tiene 'internal' cuando operator_id falta
operator_missing_analysis = df1[df1['operator_id'].isnull()]['internal'].value_counts(dropna=False)
print("\nDistribución de 'internal' cuando 'operator_id' falta:")
for value, count in operator_missing_analysis.items():
    percentage = (count / df1['operator_id'].isnull().sum()) * 100
    print(f"  internal = {value}: {count:,} casos ({percentage:.1f}%)")

print("\n" + "-"*50)

# 2. ¿Qué significa esto para tu hipótesis?
print("\n ANÁLISIS DE TU HIPÓTESIS:")
internal_true_missing = operator_missing_analysis.get(True, 0)
internal_false_missing = operator_missing_analysis.get(False, 0)
internal_nan_missing = operator_missing_analysis.get(np.nan, 0)

print(f"Llamadas internas (internal=True) sin operator_id: {internal_true_missing:,}")
print(f"Llamadas NO internas (internal=False) sin operator_id: {internal_false_missing:,}")
print(f" Casos donde internal=NaN sin operator_id: {internal_nan_missing:,}")

print(f"\n CONCLUSIÓN PRELIMINAR:")
print(f"Solo {internal_true_missing:,} de {df1['operator_id'].isnull().sum():,} casos faltantes")
print(f"son realmente llamadas internas ({(internal_true_missing/df1['operator_id'].isnull().sum()*100):.1f}%)")

print("\n" + "-"*50)

# 3. Análisis por dirección más detallado
print("\n ANÁLISIS POR DIRECCIÓN DE LLAMADA:")
direction_detail = df1[df1['operator_id'].isnull()].groupby(['direction', 'internal']).size().unstack(fill_value=0)
print("\nValores faltantes en operator_id por dirección e internal:")
print(direction_detail)
```

```python
# ===================================================================
# CÁLCULO DE MÉTRICAS PARA IDENTIFICAR OPERADORES INEFICACES
# ===================================================================

print(" CALCULANDO MÉTRICAS DE RENDIMIENTO DE OPERADORES")
print("=" * 60)

# Filtrar solo registros con operator_id válido
df_operators = df1[df1['operator_id'].notna()].copy()
print(f"Registros con operator_id válido: {len(df_operators):,}")

print("\n" + "="*50)

# MÉTRICA 1: Volumen y duración de llamadas por operador
print("\n VOLUMEN Y DURACIÓN DE LLAMADAS POR OPERADOR:")
operator_metrics = df_operators.groupby('operator_id').agg({
    'operator_id': 'count',  # Total de llamadas
    'call_duration': ['sum', 'mean', 'median']  # Duración total, promedio y mediana
}).round(2)

# Aplanar los nombres de columnas
operator_metrics.columns = ['total_llamadas', 'duracion_total', 'duracion_promedio', 'duracion_mediana']
operator_metrics = operator_metrics.reset_index()

print(f"Operadores únicos: {len(operator_metrics)}")
print(f"Promedio de llamadas por operador: {operator_metrics['total_llamadas'].mean():.1f}")
print(f"Promedio de duración total por operador: {operator_metrics['duracion_total'].mean():.1f} segundos")

print("\n" + "-"*60)
print(" TOP 10 OPERADORES POR VOLUMEN:")
top_operators = operator_metrics.sort_values('total_llamadas', ascending=False).head(10)
print(top_operators.to_string(index=False))

print("\n" + "-"*60)
print(" BOTTOM 10 OPERADORES POR VOLUMEN:")
bottom_operators = operator_metrics.sort_values('total_llamadas', ascending=True).head(10)
print(bottom_operators.to_string(index=False))

print("\n" + "-"*60)
print(" ESTADÍSTICAS GENERALES:")
print(f"Promedio del equipo - Total llamadas: {operator_metrics['total_llamadas'].mean():.1f}")
print(f"Promedio del equipo - Duración total: {operator_metrics['duracion_total'].mean():.1f} seg")
print(f"Promedio del equipo - Duración promedio por llamada: {operator_metrics['duracion_promedio'].mean():.1f} seg")
```

```python
# Análisis de patrones de llamadas por operador
pattern_analysis = df_clean_operators.groupby('operator_id').agg({
    'direction': lambda x: (x == 'in').sum(),  # Llamadas entrantes
    'operator_id': 'count'  # Total llamadas
}).rename(columns={'direction': 'llamadas_entrantes', 'operator_id': 'total_llamadas'})

pattern_analysis['llamadas_salientes'] = pattern_analysis['total_llamadas'] - pattern_analysis['llamadas_entrantes']
pattern_analysis['ratio_entrantes'] = pattern_analysis['llamadas_entrantes'] / pattern_analysis['total_llamadas']
pattern_analysis['ratio_salientes'] = pattern_analysis['llamadas_salientes'] / pattern_analysis['total_llamadas']

# Mostrar los resultados del análisis de patrones
print(" ANÁLISIS DE PATRONES: ROLES DE OPERADORES")
print("=" * 55)

print(f"\nOperadores analizados (sin bot): {len(pattern_analysis)}")
print(f"Total de llamadas analizadas: {pattern_analysis['total_llamadas'].sum():,}")

print("\n ESTADÍSTICAS GENERALES:")
print(f"Promedio llamadas por operador: {pattern_analysis['total_llamadas'].mean():.1f}")
print(f"Promedio ratio entrantes: {pattern_analysis['ratio_entrantes'].mean():.2f}")
print(f"Promedio ratio salientes: {pattern_analysis['ratio_salientes'].mean():.2f}")

# Mostrar primeros 10 resultados
print("\n PRIMEROS 10 OPERADORES:")
print(pattern_analysis.head(10))
```

```python
# ===================================================================
# SISTEMA COMPLETO DE CLASIFICACIÓN DE DESEMPEÑO
# ===================================================================
print(" IMPLEMENTANDO SISTEMA DE CLASIFICACIÓN DE DESEMPEÑO")
print("=" * 65)

# PASO 1: Calcular métricas adicionales necesarias
print("\n CALCULANDO MÉTRICAS ADICIONALES...")

# Crear dataset específico para análisis de performance
df_performance = df1[
    (df1['operator_id'].notna()) &  # Solo operadores válidos
    (df1['operator_id'] != 885876.0)  # Excluir el bot que identificaste
].copy()

# Agregar métricas de tiempo de espera y tasa de éxito
performance_metrics_extended = df_performance.groupby('operator_id').agg({
    # Métricas básicas
    'operator_id': 'count',
    'is_missed_call': 'sum',
    'call_duration': ['mean', 'sum', lambda x: (x > 600).sum()],
    'direction': [
        lambda x: (x == 'in').sum(),
        lambda x: (x == 'out').sum()
    ]
}).round(2)

# Aplanar nombres de columnas
performance_metrics_extended.columns = [
    'total_llamadas', 'llamadas_perdidas', 'duracion_promedio', 'duracion_total',
    'llamadas_largas', 'llamadas_entrantes', 'llamadas_salientes'
]

performance_metrics_extended = performance_metrics_extended.reset_index()

# Calcular llamadas salientes exitosas por separado
salientes_exitosas = df_performance[
    (df_performance['direction'] == 'out') & 
    (df_performance['call_duration'] > 0)
].groupby('operator_id').size()

# Agregar la columna de salientes exitosas
performance_metrics_extended['salientes_exitosas'] = performance_metrics_extended['operator_id'].map(salientes_exitosas).fillna(0)

print(" Métricas extendidas calculadas")
print(f"Total operadores analizados: {len(performance_metrics_extended)}")
```

```python
# ===================================================================
# PASO 2: CALCULAR RATIOS Y MÉTRICAS DE RENDIMIENTO
# ===================================================================
print("\n CALCULANDO RATIOS DE RENDIMIENTO...")

# Calcular ratios y métricas adicionales
performance_metrics_extended['ratio_llamadas_perdidas'] = (
    performance_metrics_extended['llamadas_perdidas'] / 
    performance_metrics_extended['total_llamadas'] * 100
).round(2)

performance_metrics_extended['ratio_salientes_exitosas'] = (
    performance_metrics_extended['salientes_exitosas'] / 
    performance_metrics_extended['llamadas_salientes'] * 100
).fillna(0).round(2)

# Calcular tiempo promedio de espera (solo para llamadas entrantes)
tiempo_espera = df_performance[df_performance['direction'] == 'in'].groupby('operator_id').agg({
    'call_duration': 'mean'
}).rename(columns={'call_duration': 'tiempo_espera_promedio'})

# Unir con las métricas principales
performance_metrics_extended = performance_metrics_extended.merge(
    tiempo_espera, 
    left_on='operator_id', 
    right_index=True, 
    how='left'
)

performance_metrics_extended['tiempo_espera_promedio'] = performance_metrics_extended['tiempo_espera_promedio'].fillna(0).round(2)

print(" Ratios calculados exitosamente")
print(f"Columnas disponibles: {list(performance_metrics_extended.columns)}")

# Mostrar primeros resultados
print("\n PRIMEROS 5 OPERADORES CON MÉTRICAS COMPLETAS:")
print(performance_metrics_extended.head())
```

```python
# ===================================================================
# PASO 3: DEFINIR CRITERIOS DE CLASIFICACIÓN DE RENDIMIENTO
# ===================================================================
print("\n DEFINIENDO CRITERIOS DE CLASIFICACIÓN...")

# Calcular percentiles para establecer umbrales
print("\n CALCULANDO PERCENTILES PARA UMBRALES:")

# Estadísticas clave para definir umbrales
stats_summary = {
    'total_llamadas': performance_metrics_extended['total_llamadas'].describe(),
    'ratio_llamadas_perdidas': performance_metrics_extended['ratio_llamadas_perdidas'].describe(),
    'ratio_salientes_exitosas': performance_metrics_extended['ratio_salientes_exitosas'].describe(),
    'duracion_promedio': performance_metrics_extended['duracion_promedio'].describe(),
    'tiempo_espera_promedio': performance_metrics_extended['tiempo_espera_promedio'].describe()
}

for metric, stats in stats_summary.items():
    print(f"\n{metric.upper()}:")
    print(f"  Promedio: {stats['mean']:.2f}")
    print(f"  Mediana: {stats['50%']:.2f}")
    print(f"  P25: {stats['25%']:.2f}")
    print(f"  P75: {stats['75%']:.2f}")

print("\n" + "="*60)
```

```python
# ===================================================================
# PASO 4: DEFINIR UMBRALES ESPECÍFICOS DE CLASIFICACIÓN
# ===================================================================
print("\n DEFINIENDO UMBRALES DE CLASIFICACIÓN BASADOS EN PERCENTILES...")

# Definir umbrales basados en los percentiles calculados
umbrales = {
    # Volumen de llamadas (más llamadas = mejor)
    'volumen_alto': performance_metrics_extended['total_llamadas'].quantile(0.75),  # P75 = 54
    'volumen_bajo': performance_metrics_extended['total_llamadas'].quantile(0.25),  # P25 = 6
    
    # Ratio de llamadas perdidas (menos perdidas = mejor)
    'perdidas_alto': performance_metrics_extended['ratio_llamadas_perdidas'].quantile(0.75),  # P75 = 46.15%
    'perdidas_bajo': performance_metrics_extended['ratio_llamadas_perdidas'].quantile(0.25),  # P25 = 12.50%
    
    # Ratio de salientes exitosas (más exitosas = mejor)
    'exitosas_alto': performance_metrics_extended['ratio_salientes_exitosas'].quantile(0.75),  # P75 = 57.14%
    'exitosas_bajo': performance_metrics_extended['ratio_salientes_exitosas'].quantile(0.25),  # P25 = 0%
    
    # Duración promedio (rango óptimo)
    'duracion_optima_min': 120,  # 2 minutos mínimo
    'duracion_optima_max': 900,  # 15 minutos máximo
    
    # Tiempo de espera (menos espera = mejor)
    'espera_alto': performance_metrics_extended['tiempo_espera_promedio'].quantile(0.75),  # P75 = 274.82
    'espera_bajo': performance_metrics_extended['tiempo_espera_promedio'].quantile(0.25)   # P25 = 0
}

print("\n UMBRALES DEFINIDOS:")
for criterio, valor in umbrales.items():
    print(f"  {criterio}: {valor:.2f}")

print("\n" + "="*60)
```

```python
# PASO 5: NORMALIZACIÓN DE MÉTRICAS PARA CLASIFICACIÓN
print("\n NORMALIZANDO MÉTRICAS PARA CLASIFICACIÓN...")

# Crear copia para trabajar
df_scoring = performance_metrics_extended.copy()

# 1. VOLUMEN DE LLAMADAS (30%) - Más llamadas = mejor score
volumen_max = df_scoring['total_llamadas'].max()
volumen_min = df_scoring['total_llamadas'].min()
df_scoring['score_volumen'] = ((df_scoring['total_llamadas'] - volumen_min) / 
                               (volumen_max - volumen_min) * 100).round(2)

# 2. RATIO LLAMADAS PERDIDAS (20%) - Menos perdidas = mejor score
# Invertir la escala: 0% perdidas = 100 puntos, 100% perdidas = 0 puntos
df_scoring['score_perdidas'] = (100 - df_scoring['ratio_llamadas_perdidas']).clip(0, 100).round(2)

# 3. RATIO SALIENTES EXITOSAS (25%) - Más exitosas = mejor score
df_scoring['score_exitosas'] = df_scoring['ratio_salientes_exitosas'].round(2)

print(" Normalización completada")
```

```python
# 5.1 DURACIÓN PROMEDIO (25%) - Rango óptimo: 5 min - 1 hora
def calcular_score_duracion(duracion):
    """
    Calcula score de duración basado en rango óptimo
    - Muy corta (< 300 seg): penalización
    - Óptima (300-3600 seg): score máximo
    - Muy larga (> 3600 seg): penalización
    """
    if duracion < 300:  # Menos de 5 minutos
        return (duracion / 300) * 80  # Máximo 80 puntos
    elif duracion <= 3600:  # Entre 5 min y 1 hora
        return 100  # Puntuación máxima
    else:  # Más de 1 hora
        # Penalización progresiva después de 1 hora
        exceso = duracion - 3600
        penalizacion = min(exceso / 3600 * 50, 50)  # Máximo 50 puntos de penalización
        return max(100 - penalizacion, 20)  # Mínimo 20 puntos

# Aplicar la función
df_scoring['score_duracion'] = df_scoring['duracion_promedio'].apply(calcular_score_duracion).round(2)

print(" Score de duración calculado")
print("\nPrimeros 5 operadores con todos los scores:")
print(df_scoring[['operator_id', 'score_volumen', 'score_perdidas', 'score_exitosas', 'score_duracion']].head())
```

```python
# SISTEMA COMPLETO: SCORE FINAL + CLASIFICACIÓN + VALIDACIÓN
# ===================================================================
print(" IMPLEMENTANDO SISTEMA COMPLETO DE CLASIFICACIÓN")
print("=" * 60)

# PASO 1: CALCULAR SCORE FINAL PONDERADO
print("\n CALCULANDO SCORE FINAL PONDERADO...")
print("Pesos definidos:")
print("  • Volumen: 30%")
print("  • Llamadas perdidas: 20%") 
print("  • Salientes exitosas: 25%")
print("  • Duración: 25%")

# Calcular score final con pesos
df_scoring['score_final'] = (
    df_scoring['score_volumen'] * 0.30 +
    df_scoring['score_perdidas'] * 0.20 +
    df_scoring['score_exitosas'] * 0.25 +
    df_scoring['score_duracion'] * 0.25
).round(2)

print(" Score final calculado")

# PASO 2: IMPLEMENTAR CATEGORÍAS DE CLASIFICACIÓN
print("\n IMPLEMENTANDO CATEGORÍAS DE CLASIFICACIÓN...")

def clasificar_operador(score):
    """
    Clasifica operadores según su score final
    """
    if score >= 80:
        return "Excelente"
    elif score >= 65:
        return "Bueno"
    elif score >= 50:
        return "Regular"
    else:
        return "Deficiente"

# Aplicar clasificación
df_scoring['clasificacion'] = df_scoring['score_final'].apply(clasificar_operador)

print(" Clasificación aplicada")
print("\n Criterios de clasificación:")
print("  • Excelente: ≥ 80 puntos")
print("  • Bueno: 65-79 puntos")
print("  • Regular: 50-64 puntos")
print("  • Deficiente: < 50 puntos")

# PASO 3: ANÁLISIS DE DISTRIBUCIÓN
print("\n📈 DISTRIBUCIÓN DE CLASIFICACIONES:")
distribucion = df_scoring['clasificacion'].value_counts()
total_operadores = len(df_scoring)

for categoria, cantidad in distribucion.items():
    porcentaje = (cantidad / total_operadores) * 100
    print(f"  • {categoria}: {cantidad} operadores ({porcentaje:.1f}%)")

# PASO 4: VALIDACIÓN CON CASOS ESPECÍFICOS
print("\n🔍 VALIDACIÓN CON CASOS ESPECÍFICOS:")
print("\n" + "="*70)

# Top 5 mejores operadores
print("🥇 TOP 5 MEJORES OPERADORES:")
top_5 = df_scoring.nlargest(5, 'score_final')[
    ['operator_id', 'total_llamadas', 'ratio_llamadas_perdidas', 
     'ratio_salientes_exitosas', 'duracion_promedio', 'score_final', 'clasificacion']
]
print(top_5.to_string(index=False))

print("\n" + "-"*70)


# Bottom 5 operadores
print("🔻 BOTTOM 5 OPERADORES:")
bottom_5 = df_scoring.nsmallest(5, 'score_final')[
    ['operator_id', 'total_llamadas', 'ratio_llamadas_perdidas', 
     'ratio_salientes_exitosas', 'duracion_promedio', 'score_final', 'clasificacion']
]
print(bottom_5.to_string(index=False))

print("\n" + "="*70)

# PASO 5: ANÁLISIS DETALLADO DE VALIDACIÓN
print("\n📋 ANÁLISIS DETALLADO DE VALIDACIÓN:")

# Validar un operador excelente
operador_excelente = df_scoring[df_scoring['clasificacion'] == 'Excelente'].iloc[0]
print(f"\n EJEMPLO OPERADOR EXCELENTE (ID: {operador_excelente['operator_id']}):")
print(f"   Score final: {operador_excelente['score_final']}")
print(f"   Llamadas totales: {operador_excelente['total_llamadas']}")
print(f"   Ratio perdidas: {operador_excelente['ratio_llamadas_perdidas']:.3f}")
print(f"   Ratio exitosas: {operador_excelente['ratio_salientes_exitosas']:.3f}")
print(f"   Duración promedio: {operador_excelente['duracion_promedio']:.1f} min")

# Validar un operador deficiente
if len(df_scoring[df_scoring['clasificacion'] == 'Deficiente']) > 0:
    operador_deficiente = df_scoring[df_scoring['clasificacion'] == 'Deficiente'].iloc[0]
    print(f"\n EJEMPLO OPERADOR DEFICIENTE (ID: {operador_deficiente['operator_id']}):")
    print(f"   Score final: {operador_deficiente['score_final']}")
    print(f"   Llamadas totales: {operador_deficiente['total_llamadas']}")
    print(f"   Ratio perdidas: {operador_deficiente['ratio_llamadas_perdidas']:.3f}")
    print(f"   Ratio exitosas: {operador_deficiente['ratio_salientes_exitosas']:.3f}")
    print(f"   Duración promedio: {operador_deficiente['duracion_promedio']:.1f} min")

print("\n RESUMEN FINAL DEL SISTEMA:")
print(f"   Total operadores evaluados: {len(df_scoring)}")
print(f"   Score promedio: {df_scoring['score_final'].mean():.2f}")
print(f"   Desviación estándar: {df_scoring['score_final'].std():.2f}")
print("\n SISTEMA DE CLASIFICACIÓN COMPLETADO")
```

```python
# ===================================================================
# PREPARACIÓN PARA PRUEBA T DE STUDENT
# ===================================================================
print("\n PREPARACIÓN PARA PRUEBA ESTADÍSTICA")
print("=" * 50)

# Definir los grupos para la comparación
grupo_deficiente = df_scoring[df_scoring['clasificacion'] == 'Deficiente']['score_final']
grupo_resto = df_scoring[df_scoring['clasificacion'] != 'Deficiente']['score_final']

print(f"\n TAMAÑOS DE GRUPOS:")
print(f"Operadores Deficientes: {len(grupo_deficiente)}")
print(f"Operadores No Deficientes: {len(grupo_resto)}")

print(f"\n ESTADÍSTICAS DESCRIPTIVAS:")
print(f"\nGRUPO DEFICIENTE:")
print(f"  Media: {grupo_deficiente.mean():.2f}")
print(f"  Desviación estándar: {grupo_deficiente.std():.2f}")
print(f"  Mediana: {grupo_deficiente.median():.2f}")

print(f"\nGRUPO NO DEFICIENTE:")
print(f"  Media: {grupo_resto.mean():.2f}")
print(f"  Desviación estándar: {grupo_resto.std():.2f}")
print(f"  Mediana: {grupo_resto.median():.2f}")

print(f"\n DIFERENCIA DE MEDIAS:")
diferencia_medias = grupo_resto.mean() - grupo_deficiente.mean()
print(f"Diferencia (No Deficiente - Deficiente): {diferencia_medias:.2f} puntos")
```

```python
# ===================================================================
# PRUEBA T DE STUDENT DE UNA COLA
# ===================================================================
print(" IMPLEMENTANDO PRUEBA T DE STUDENT DE UNA COLA")
print("=" * 60)

# Importar librerías necesarias
from scipy import stats
import numpy as np

# Definir nivel de significancia
alpha = 0.05
print(f"\n CONFIGURACIÓN DE LA PRUEBA:")
print(f"Nivel de significancia (α): {alpha}")
print(f"Tipo de prueba: Una cola (unilateral)")

# Hipótesis
print(f"\n HIPÓTESIS:")
print(f"H₀: μ_deficiente = μ_no_deficiente")
print(f"H₁: μ_deficiente < μ_no_deficiente")

# Verificar supuestos antes de la prueba
print(f"\n VERIFICACIÓN DE SUPUESTOS:")
print(f"Tamaño grupo deficiente: {len(grupo_deficiente)}")
print(f"Tamaño grupo no deficiente: {len(grupo_resto)}")
print(f"Ambos grupos > 30: {'✅' if len(grupo_deficiente) > 30 and len(grupo_resto) > 30 else '❌'}")

# Realizar la prueba t de dos muestras independientes
t_statistic, p_value_bilateral = stats.ttest_ind(grupo_deficiente, grupo_resto, equal_var=False)

# Para prueba de una cola, dividir p-valor entre 2
p_value_unilateral = p_value_bilateral / 2

print(f"\n RESULTADOS DE LA PRUEBA:")
print(f"Estadístico t: {t_statistic:.4f}")
print(f"P-valor (bilateral): {p_value_bilateral:.6f}")
print(f"P-valor (unilateral): {p_value_unilateral:.6f}")
```

```python
# ===================================================================
# INTERPRETACIÓN COMPLETA DE LA PRUEBA T DE UNA COLA
# ===================================================================
print(" INTERPRETACIÓN DE RESULTADOS")
print("=" * 50)

# Resultados obtenidos
print(f"\n RESULTADOS OBTENIDOS:")
print(f"Estadístico t: {t_statistic:.4f}")
print(f"P-valor (unilateral): {p_value_unilateral:.6f}")
print(f"Nivel de significancia: {alpha}")

# Decisión estadística
print(f"\n DECISIÓN ESTADÍSTICA:")
if p_value_unilateral < alpha:
    decision = "RECHAZAR H₀"
    conclusion = "ACEPTAR H₁"
else:
    decision = "NO RECHAZAR H₀"
    conclusion = "NO HAY EVIDENCIA SUFICIENTE"

print(f"P-valor ({p_value_unilateral:.6f}) {'<' if p_value_unilateral < alpha else '>='} α ({alpha})")
print(f"Decisión: {decision}")
print(f"Conclusión: {conclusion}")

# Interpretación práctica
print(f"\n INTERPRETACIÓN PRÁCTICA:")
if p_value_unilateral < alpha:
    print("✅ EVIDENCIA ESTADÍSTICAMENTE SIGNIFICATIVA")
    print(f"   • Los operadores deficientes tienen un score promedio")
    print(f"     significativamente MENOR que los no deficientes")
    print(f"   • Diferencia de medias: {diferencia_medias:.2f} puntos")
    print(f"   • Esta diferencia NO es debida al azar")
    print(f"   • Confianza: {(1-alpha)*100}%")
else:
    print("❌ NO HAY EVIDENCIA ESTADÍSTICAMENTE SIGNIFICATIVA")

# Tamaño del efecto
print(f"\n📏 TAMAÑO DEL EFECTO:")
# Calcular Cohen's d
pooled_std = np.sqrt(((len(grupo_deficiente)-1)*grupo_deficiente.var() + 
                     (len(grupo_resto)-1)*grupo_resto.var()) / 
                    (len(grupo_deficiente) + len(grupo_resto) - 2))
cohens_d = diferencia_medias / pooled_std
print(f"Cohen's d: {cohens_d:.4f}")

if abs(cohens_d) < 0.2:
    efecto = "PEQUEÑO"
elif abs(cohens_d) < 0.8:
    efecto = "MEDIANO"
else:
    efecto = "GRANDE"
    
print(f"Tamaño del efecto: {efecto}")
```
